```python
# step 6 : how to handel the missing values
# syntex

df2 = pd.DataFrame(
    {
        'name': [np.nan,'B',np.nan,'A'],
        'roll_no':[np.nan,2,'None',1],
        'marks':[np.nan,60,70,20]

    })
display(df2)
```

```python
# 1 deleting all the nan values here we use 2 methods 'ANY' and 'ALL'
#syntax: df.dropna(inplace = True)

df2['name']
```

```python
df2['name'].dropna(inplace=True)
```

```python
df2['name']
```

```python
df2['marks'].dropna(inplace=True)
```

```python
df2['marks'] #here we are not droping any values , to overcome that we use next up
```

```python
df2
```

```python
# if you want to drop the entire row if all the values are NaN

df2.dropna(how='all',inplace=True)  #it will remove entries row if the value in NaN
df2
```

```python
# if you want to delete row oif it has at least one NaN in it we use
df2.dropna(how='any',inplace=True)  #it will remove entries row if the value in NaN
df2
```

```python
# 2 filling nan value either with 0 or mean or median

df4 = pd.DataFrame(
    {
    'name':['A','B',np.nan,'A',np.nan,'C'],
    'roll_no':[1,2,'None',1,np.nan,3], #'None' is in str formate thats why it is not NONE ,now it is srt
    'marks':[np.nan,60,70,20,40,50]
    })
df4
```

```python
# np.nan are the real missing values
#forward fill (ffill) ---- replace NaN with the previous not-null value in the column value, means in name col 2 row it will occupy B in place af Nan
```

```python
df4['name']
```

```python
df4['name'].ffill()
```

```python
#backward fill ---> replaces NaN with the next non-null
# it will work under cyclic manner if last value will be replaced by first value only when last cell is Nan
```

```python
df4['name']
```

```python
df4['name'].bfill()
```

```python
#mean()
a=df4['marks'].mean()
a
```

```python
df4['marks'].fillna(a) #fillna use to assign the value for missing data with mean value which is stored in a variable
```

```python
df4['marks'].fillna(b)  #fillna use to assign the value for missing data with median  value which is stored in b variable
```
