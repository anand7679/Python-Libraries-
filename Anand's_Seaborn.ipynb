```python
# upper bound(UB)---the maximum acceptable value before data is considered too high
# syntax: Q1 + 1.5 *IQR

# lower bound(LB)---the minimum acceptable value before data is considered too low
# syntax: Q1 - 1.5 *IQR

# outliers------the data points that lie b/w the LB,UP

# winsorization / capping -------- replace the extreme value with nearest bound
```

```python
# calculate Q1,Q3 and IQR
Q1=df1['alcohol'].quantile(0.25) #lowerbound
Q3=df1['alcohol'].quantile(0.75)  #upperbound
IQR=Q3-Q1   #IQR --> interQuartile range >> it is the measure of how spread out the middle 50% of your data
print(Q1,Q3,IQR)
```

```python
# calculate UB and LB

upper_bound=Q3+1.5*IQR
lower_bound=Q1-1.5*IQR
print(lower_bound,upper_bound)
```

```python
# calculate outlier(s

df1[(df1['alcohol']>upper_bound) |  (df1['alcohol']<lower_bound)]
```

```python
# winsorization / capping

sakshi= df1['alcohol'].clip(lower=lower_bound,upper=upper_bound)
sakshi
```

```python
# CO-RELATION MATRIX PLOT
#select only numerical columns for matrix plot

a=df1.select_dtypes(include=['float64','Int64']) #select_dtype--> use to extract user require datatype
a
```

```python
# calculate CO-RELATION MATRIX
# here we use heatmap keyword to define in sns.heatmap
matrix = a.corr() #corr --> built in funt
matrix
```

```python
# plot heatmap
plt.figure(figsize=(12,10))
sns.heatmap(matrix,
            annot=True, # marks correlation values of eact points
            fmt='0.2f', # it will add the decimal values
            cmap='coolwarm',
            linewidth=5,
            linecolor='black',

            )
plt.show()
```
