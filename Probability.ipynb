```python
# Roll of two daice
import itertools # used to calculate the sample space
dise={1,2,3,4,5,6}
sample_space=list(itertools.product(dise,repeat=2))
len(sample_space)
```

```python
# sum of dice ==7
event=[i for i in sample_space if sum(i)==7]
print(event, len(event))
```

```python
# TOssing two coins
coin={"H","T"}
sample_space=list(itertools.product(coin,repeat=2))
sample_space
```

```python
# Event where both coin shows Heads
event=[i for i in sample_space if i==("H","H")]
print(event, len(event))
```

```python
# Distribution >>>
# 1.Binamial Distribution -->> the number of successes n independent binomial trials

from scipy.stats import binom # importing the binomial

n = 3
p = 1/8
k = 3 # k  means number intervals
success = binom.pmf(k,n,p) #A Probability Mass Function (PMF) gives the probability that a discrete random variable is exactly equal to some value.

success

```

```python
# 2.poisson distribution -->> the no. of events in a fixed time of intervel,here the complete task will be distributed and it will make task easier

from scipy.stats import poisson
# ex 1  call center , average call per hour = 15
# the person did 7 calls per hour so lets find probability

target =15
complete = 7
success =poisson.pmf(complete,target)
success
```

```python
# 3.Geo-metric distribution -->> number of trils before first success ,here it will calculate the probability before geting the 1 st succes

from scipy.stats import geom

all_head=1/8
target=5
success =geom.pmf(target,all_head)
success


# remaining distribution are covered in statistics
```
